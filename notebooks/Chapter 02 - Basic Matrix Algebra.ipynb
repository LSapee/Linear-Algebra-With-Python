# <font face="gotham" color="purple">행렬 연산</font>

행렬 _덧셈_에는 다음 성질이 있습니다:
1. $A+ B= B+ A$
2. $(A+B)+ C=A+(B+C)$
3. $c(A+B)=cA+cB$
4. $(c+d)A=cA+c{D}$
5. $c(dA)=(cd)A$
6. $A+{0}=A$, 여기서 ${0}$은 영행렬입니다
7. 모든 $A$에는 $A+(-A)=0$을 만족하는 덧셈의 역원 $-A$가 있습니다.

성질이 직관적이므로 증명은 생략합니다. 행렬 _곱셈_에는 다음 성질이 있습니다:
1. $ A({BC})=({AB}) C$
2. $c({AB})=(cA)B=A(cB)$
3. $A(B+ C)={AB}+{AC}$
4. $(B+C)A={BA}+{CA}$

두 가지 곱셈을 구분해야 합니다. $A\odot B$로 쓰는 _아다마르 곱_은 같은 위치의 원소끼리 곱하고, $AB$로 쓰는 _행렬 곱_은 행과 열을 이용해 계산합니다.

```python
A = np.array([[1, 2], [3, 4]])  # 예제에 사용할 첫 번째 행렬 A를 만듭니다.
B = np.array([[5, 6], [7, 8]])  # 두 번째 행렬 B를 만듭니다.
```

```python
A*B  # 같은 위치의 원소끼리 곱하는 아다마르 곱을 계산합니다.
```

```python
A@B  # 행과 열의 내적으로 계산하는 행렬 곱을 구합니다.
```

행렬 곱의 각 원소가 어떻게 계산되는지 직접 확인해 봅시다.

```python
np.sum(A[0,:]*B[:,0])  # 곱의 첫째 행 첫째 열 원소를 계산합니다.
np.sum(A[1,:]*B[:,0])  # 둘째 행 첫째 열 원소를 계산합니다.
np.sum(A[0,:]*B[:,1])  # 첫째 행 둘째 열 원소를 계산합니다.
np.sum(A[1,:]*B[:,1])  # 둘째 행 둘째 열 원소를 계산합니다.
```

## <font face="gotham" color="purple">행렬 곱의 교환법칙</font>

행렬 곱에는 일반적으로 교환법칙이 성립하지 않습니다. 즉, $AB\neq BA$입니다. 다음 두 행렬로 확인해 봅시다.

```python
A = sy.Matrix([[3, 4], [7, 8]])  # 수치가 주어진 행렬 A를 만듭니다.
B = sy.Matrix([[5, 3], [2, 1]])  # 비교할 행렬 B를 만듭니다.
A*B  # AB를 계산합니다.
B*A  # BA를 계산해 순서에 따른 차이를 확인합니다.
```

어떤 행렬끼리 곱셈 순서를 바꿔도 결과가 같을까요? 기호 계산으로 조건을 찾아봅시다.

```python
A = sy.Matrix([[a, b], [c, d]])  # 기호를 원소로 갖는 행렬 A를 만듭니다.
B = sy.Matrix([[e, f], [g, h]])  # 기호를 원소로 갖는 행렬 B를 만듭니다.
A*B  # AB의 각 원소를 계산합니다.
B*A  # BA의 각 원소를 계산합니다.
```

$AB=BA$인지 확인하려면 $AB-BA=0$을 만족하는지 살펴보면 됩니다.

```python
M = A*B - B*A; M  # AB와 BA의 차를 구해 표시합니다.
```

이 조건을 원소별로 쓰면 다음 연립일차방정식을 얻습니다.
$$
\begin{align}
b g - c f&=0 \\
 a f - b e + b h - d f&=0\\
- a g + c e - c h + d g&=0 \\
- b g + c f&=0
\end{align}
$$

$a,b,c,d$를 계수로 취급하면 가우스-조던 소거법을 적용할 수 있습니다. 각 식을 $e,f,g,h$에 대해 정리해 확대행렬을 만들겠습니다.
$$
\begin{align*}
b g-c f=0 \quad &\Rightarrow-c f+b g+0 e+0 h=0\\
a f-b e+b h-d f=0 \quad &\Rightarrow(a-d) f+0 g-b e+b h=0\\
-a g+c e-c h+d g=0 \quad &\Rightarrow 0 f+(d-a) g+c e-c h=0\\
-b g+c f=0 \quad &\Rightarrow c f-b g+0 e+0 h=0
\end{align*}
$$

따라서 확대행렬은 다음과 같습니다.
$$
\begin{equation}
\left[\begin{array}{cccc:c}
-c & b & 0 & 0 & 0 \\
a-d & 0 & -b & b & 0 \\
0 & d-a & c & -c & 0 \\
c & -b & 0 & 0 & 0
\end{array}\right]
\end{equation}
$$

```python
A_aug = sy.Matrix([[-c, b, 0, 0], [a-d,0, -b, b], [0, d-a, c, -c], [c, -b, 0, 0]]); A_aug  # 교환 조건을 연립방정식의 확대행렬로 나타냅니다.
```

가우스-조던 소거법으로 행렬을 기약 행 사다리꼴로 바꿉니다.

```python
A_aug.rref()  # 확대행렬을 기약 행 사다리꼴로 바꿉니다.
```

일반해는 다음과 같이 표현됩니다.
$$
\begin{equation}
\left(\begin{array}{l}
e \\
f \\
g \\
h
\end{array}\right)=c_1\left(\begin{array}{c}
\frac{b}{a-d} \\
\frac{b c}{a b-b d} \\
1 \\
0
\end{array}\right)+c_2\left(\begin{array}{c}
-\frac{b}{a-d} \\
-\frac{c}{a-d} \\
0 \\
1
\end{array}\right)
\end{equation}
$$

```python
import sympy as sp  # 기호 계산에 사용할 SymPy를 불러옵니다.

a, b, c, d = sp.symbols('a b c d')  # 행렬 A의 네 원소를 기호로 정의합니다.

e, f, g, h = sp.symbols('e f g h')  # 행렬 B의 네 원소를 기호로 정의합니다.

A = sp.Matrix([[a, b], [c, d]])  # 기호로 이루어진 행렬 A를 만듭니다.
B = sp.Matrix([[e, f], [g, h]])  # 기호로 이루어진 행렬 B를 만듭니다.

AB = A * B  # 두 행렬의 곱 AB를 계산합니다.
BA = B * A  # 곱셈 순서를 바꾼 BA를 계산합니다.

equations = [  # AB=BA 조건의 식 목록을 시작합니다.
    sp.Eq(AB[i, j], BA[i, j]) for i in range(2) for j in range(2)  # 각 위치의 원소가 같은지 나타내는 식을 만듭니다.
]  # 식 목록을 닫습니다.

solution = sp.solve(equations, (e, f, g, h))  # 식에서 B의 원소에 대한 해를 구합니다.

print("Solution for B elements:")  # 해를 출력하기 위한 제목을 표시합니다.
for sol in solution:  # 각 해를 순서대로 확인합니다.
    print(f"{sol}: {solution[sol]}")  # 현재 기호의 해를 출력합니다.

c1, c2 = sp.symbols('c1 c2')  # 일반해를 나타낼 두 매개변수를 정의합니다.

parametric_solution = {  # 매개변수로 표현할 해의 사전을 시작합니다.
    e: solution[e].subs({g: c1, h: c2}),  # e에 대한 해에 매개변수를 대입합니다.
    f: solution[f].subs({g: c1, h: c2}),  # f에 대한 해에 매개변수를 대입합니다.
    g: c1,  # g를 첫 번째 매개변수로 둡니다.
    h: c2  # h를 두 번째 매개변수로 둡니다.
}  # 매개변수 해의 사전을 닫습니다.

print("\nParameterized solution in LaTeX format:")  # LaTeX 형식 출력의 제목을 표시합니다.
for sol in parametric_solution:  # 각 기호의 매개변수 해를 확인합니다.
    print(f"{sp.latex(sol)}: {sp.latex(parametric_solution[sol])}")  # 기호와 해를 LaTeX 문자열로 출력합니다.
```

$$
e: c_{2} + \frac{c_{1} \left(a - d\right)}{c}\\
f: \frac{b c_{1}}{c}\\
g: c_{1}\\
h: c_{2}\\
$$

# <font face="gotham" color="purple">전치행렬</font>

$n\times m$ 행렬 $A$의 전치행렬 $A^T$는 $m\times n$ 행렬입니다.

```python
A = sy.Matrix([[1, 2, 3], [4, 5, 6]]); A
A.transpose()
```

전치행렬에는 다음 성질이 있습니다.
1. $(A^T)^T$
2. $(A+B)^T=A^T+B^T$
3. $(cA)^T=cA^T$
4. $(AB)^T=B^TA^T$

마지막 성질을 SymPy로 확인해 봅시다. 먼저 $A$와 $B$를 곱한 뒤 결과를 전치해 $(AB)^T$를 구합니다.

```python
A = sy.Matrix([[a, b], [c, d], [e, f]])
B = sy.Matrix([[g, h, i], [j, k, l]])
AB = A*B
AB_t = AB.transpose(); AB_t
```

이번에는 $A$와 $B$를 각각 전치한 다음 반대 순서로 곱해 $B^TA^T$를 구합니다.

```python
B_t_A_t = B.transpose()*A.transpose()
B_t_A_t
```

두 결과가 같은지 확인합니다.

```python
AB_t == B_t_A_t
```

# <font face="gotham" color="purple">기본행렬</font>

단위행렬에 기본 행 연산을 한 번 적용해 얻는 행렬을 **기본행렬**이라고 합니다. 예를 들어 다음과 같습니다.

$$
\left[
\begin{array}{ccc}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{array}
\right]
\begin{array}{c}
R_1 \leftrightarrow R_2 \\
~ \\
~
\end{array}
\qquad \Longrightarrow \qquad
\left[
\begin{array}{ccc}
0 & 1 & 0 \\
1 & 0 & 0 \\
0 & 0 & 1
\end{array}
\right]
$$

$R_1\leftrightarrow R_2$는 첫째 행과 둘째 행을 바꾸라는 뜻입니다. 이 연산으로 얻은 기본행렬을 $E$라 하면 $EA$는 $A$에 똑같은 행 교환을 적용한 결과입니다.

먼저 행렬 $A$를 만듭니다.

```python
A = sy.randMatrix(3, percent = 80); A # generate a random matrix with 80% of entries being nonzero
```

$R_1\leftrightarrow R_2$에 해당하는 기본행렬을 만듭니다.

```python
E = sy.Matrix([[0, 1, 0], [1, 0, 0], [0, 0, 1]]);E
```

$E$를 $A$의 왼쪽에서 곱하면 $A$의 첫째 행과 둘째 행도 서로 바뀝니다.

```python
E*A
```

단위행렬의 한 행에 다른 행의 배수를 더해도 기본행렬을 얻을 수 있습니다.

$$

\left[
\begin{array}{ccc}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{array}
\right]\\

\begin{array}{c}
R_3-7R_1
\end{array}
\longrightarrow
\left[
\begin{array}{ccc}
1 & 0 & 0 \\
0 & 1 & 0 \\
-7 & 0 & 1
\end{array}
\right]

$$

SymPy로 결과를 확인해 봅시다.

```python
A = sy.randMatrix(3, percent = 80); A
E = sy.Matrix([[1, 0, 0], [0, 1, 0], [-7, 0, 1]]); E
```

$EA$를 계산하면 $A$의 셋째 행에 $-7$배의 첫째 행을 더한 결과가 나옵니다.

```python
E*A
```

$A$에 같은 행 연산을 직접 적용해도 결과가 일치하는지 확인해 봅시다.

```python
EA = sy.matrices.MatrixBase.copy(A)
EA[2,:]=-7*EA[0,:]+EA[2,:]
EA
```

다음 절에서는 가역행렬이 여러 기본행렬의 곱으로 표현될 수 있다는 중요한 성질을 살펴보겠습니다.

# <font face="gotham" color="purple">역행렬</font>

$AB=BA=\mathbf{I}$를 만족하는 $B$를 $A$의 역행렬이라고 하며, $B=A^{-1}$로 씁니다.

NumPy의 `np.linalg.inv()`로 역행렬을 계산할 수 있습니다. 먼저 행렬 $A$를 만들어 봅시다.

```python
A = np.round(10*np.random.randn(5,5)); A
```

```python
Ainv = np.linalg.inv(A); Ainv
```

계산한 두 행렬이 실제로 서로의 역행렬인지 확인합니다.

```python
A@Ainv
```

출력에 보이는 `-0.`은 반올림 때문에 아주 작은 음수가 0처럼 표시된 것입니다.

## <font face="gotham" color="purple">가우스-조던 소거법으로 역행렬 구하기</font>

역행렬을 구하는 한 가지 방법은 확대행렬 $[A\,|\,\mathbf{I}]$를 만드는 것입니다. 기본 행 연산을 반복해 왼쪽의 $A$를 기약 행 사다리꼴로 바꿉니다. $A$가 가역행렬이라면 왼쪽은 단위행렬이 되고, 오른쪽에는 $A^{-1}$가 남습니다.

이 과정을 SymPy의 `rref()`로 확인해 봅시다.

```python
AI = np.hstack((A, I)) # stack the matrix A and I horizontally
AI = sy.Matrix(AI); AI
```

```python
AI_rref = AI.rref(); AI_rref
```

결과 행렬의 오른쪽 부분이 $A^{-1}$입니다.

```python
Ainv = AI_rref[0][:,5:];Ainv # extract the RHS block
```

맨 앞에서 정의한 함수를 사용해 소수를 넷째 자리까지 표시하면 결과를 읽기 쉽습니다.

```python
round_expr(Ainv, 4) 
```

$AA^{-1}=\mathbf{I}$인지 확인해 봅시다.

```python
A = sy.Matrix(A)
round_expr(A*Ainv, 4) 
```

단위행렬이 나왔으므로, 앞에서 꺼낸 오른쪽 부분이 실제로 $A^{-1}$임을 확인했습니다.

## <font face="gotham" color="purple">예제: 역행렬이 존재하는 조건</font>

다음 행렬이 가역행렬이 되지 않게 하는 $\lambda$의 값을 구해 봅시다.
$$A=\left[ \begin{matrix}3 &\lambda &1\cr 2 & -1 & 6\cr 1 & 9 & 4\end{matrix}\right]$$


이번에도 SymPy로 조건을 구해 보겠습니다.

```python
lamb = sy.symbols('lamda') # SymPy will automatically render into LaTeX greek letters
A = np.array([[3, lamb, 1], [2, -1, 6], [1, 9, 4]])
I = np.eye(3); A
```

단위행렬을 붙여 확대행렬을 만듭니다.

```python
AI = np.hstack((A, I))
AI = sy.Matrix(AI); AI
```

```python
AI_rref = AI.rref()
AI_rref
```

$A$가 가역행렬이 되려면 계산 결과의 분모가 모두 0이 아니어야 합니다.
$$
\begin{align*}
-6\lambda -465 &\neq0\\
4 \lambda + 310 &\neq 0\\
2 \lambda + 155 &\neq 0
\end{align*}
$$
세 식은 서로 상수배 관계이므로 결국 같은 조건을 나타냅니다.

분모가 0이 되는 $\lambda$의 값을 구합니다.

```python
sy.solvers.solve(-6*lamb-465, lamb)
```

이 값에서는 $A$의 역행렬이 존재하지 않습니다. 행렬식으로도 확인해 봅시다. $|A|=0$이면 역행렬이 없으므로, 구한 $\lambda$를 $A$에 대입해 행렬식을 계산합니다. 행렬식은 뒤에서 자세히 다룹니다.

```python
A = np.array([[3, -155/2, 1], [2, -1, 6], [1, 9, 4]])
np.linalg.det(A)
```

계산 결과 $|A|=0$입니다.

따라서 $A$가 가역행렬이 되는 조건은 $\lambda\neq-\frac{155}{2}$입니다.

## <font face="gotham" color="purple">역행렬의 성질</font>

1. $A$와 $B$가 모두 가역행렬이면 $(AB)^{-1}=B^{-1}A^{-1}$.
2. $A$가 가역행렬이면 $(A^T)^{-1}=(A^{-1})^T$.
3. $A$와 $B$가 모두 가역인 대칭행렬이고 $AB=BA$이면 $A^{-1}B$ is symmetric.

_첫 번째 성질_은 다음 식으로 확인할 수 있습니다.
$$
\begin{align}
ABB^{-1}A^{-1}=AIA^{-1}=I=AB(AB)^{-1}
\end{align}
$$

_두 번째 성질_은 다음 식을 보이면 됩니다.
$$
A^T(A^{-1})^T = I
$$
행렬 곱의 전치 성질을 이용하면
$$
A^T(A^{-1})^T=(A^{-1}A)^T = I^T = I
$$

_세 번째 성질_은 다음 등식을 확인하는 것입니다.
$$
A^{-1}B = (A^{-1}B)^T
$$
다시 전치의 성질을 이용하면
$$
(A^{-1}B)^{T}=B^T(A^{-1})^T=B(A^T)^{-1}=BA^{-1}
$$
여기에 $AB=BA$라는 조건을 적용합니다.
\begin{align*}
AB&=BA\\
A^{-1}ABA^{-1}&=A^{-1}BAA^{-1}\\
BA^{-1}&=A^{-1}B
\end{align*}
이를 앞의 식에 대입하면 다음 결과를 얻습니다.
$$
(A^{-1}B)^{T}=BA^{-1}=A^{-1}B
$$
